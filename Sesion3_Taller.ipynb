```python
#Tema 6: Condicionales

total = 1100
es_miembro = True
compras = 23

# 1. Calcular envío
if total > 1000 or es_miembro:
    envio = "Gratis"
else:
    envio = "$99"

print("Costo de envío:", envio)

# 2. Calcular nivel
if compras >= 10:
    nivel = "Oro"
elif compras >= 5:
    nivel = "Plata"
else:
    nivel = "Bronce"

print("Nivel:", nivel)
```

```python
#Tema 7: BUCLE FOR
# Precios de tu compra
precios = [180, 95, 380, 30]

# 1. Imprime cada precio con un for
for precio in precios:
    print(precio)

# 2. Suma el total
total = 0

for precio in precios:
    total += precio

print("Total:", total)

# 3. ¿Cuántos cuestan más de 100?
caros = 0

for precio in precios:
    if precio > 100:
        caros += 1

print("Productos mayores a 100:", caros)
```

```python
#Tema 8: Bucle While
ahorro = 0
meta = 800000
por_mes = 20000
meses = 5

# Mientras no llegues a la meta
while ahorro < meta:
    ahorro += por_mes
    meses += 1
    print("Mes", meses, "- Ahorro acumulado:", ahorro)

# ¿Cuántos meses tardaste?
print("Meses para alcanzar la meta:", meses)
```

```python
#Tema 9: Funciones
cuenta = 650000
propina = 15      # %
amigos = 5

# 1. Regresa la cuenta más la propina
def con_propina(cuenta, porcentaje):
    return cuenta + (cuenta * porcentaje / 100)

# 2. Regresa cuánto paga cada persona
def por_persona(total, personas):
    return total / personas

# 3. Úsalas con tus datos e imprime
total = con_propina(cuenta, propina)
pago = por_persona(total, amigos)

print("Total con propina:", total)
print("Cada persona paga:", pago)
```

```python
#Tema 10: Problema Final
# Lista de precios
precios = [180, 550, 1800, 900]

# 1. Sumar el total con un for
total = 0

for precio in precios:
    total += precio

# 2. Aplicar descuento si supera 500
if total > 500:
    descuento = total * 0.10
else:
    descuento = 0

a_pagar = total - descuento

# 3. Imprimir ticket
print(f"Total: ${total}")
print(f"Descuento: ${descuento}")
print(f"A pagar: ${a_pagar}")
```
